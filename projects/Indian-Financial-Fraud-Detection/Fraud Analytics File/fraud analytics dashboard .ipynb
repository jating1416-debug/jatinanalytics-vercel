```python
import pandas as pd
import random
from faker import Faker
from dateutil.relativedelta import relativedelta

fake = Faker("en_IN")

# Customers Table Load
customers_df = pd.read_csv("Cusmtomer_data.csv")

# Card Type -> Credit Limit
card_limits = {
    "Classic": (50000, 150000),
    "Silver": (150000, 300000),
    "Gold": (300000, 700000),
    "Platinum": (700000, 1500000)
}

card_networks = [
    "Visa",
    "Mastercard",
    "RuPay",
    "American Express"
]

cards = []

card_id = 500001

for _, customer in customers_df.iterrows():

    customer_id = customer["Customer_ID"]
    segment = customer["Customer_Segment"]

    # Segment -> Card Type
    if segment == "Basic":
        first_card = "Classic"

    elif segment == "Standard":
        first_card = random.choice(["Classic", "Silver"])

    elif segment == "Gold":
        first_card = random.choice(["Silver", "Gold"])

    else:
        first_card = random.choice(["Gold", "Platinum"])

    # 1,2,3 Cards
    total_cards = random.choices(
        [1, 2, 3],
        weights=[75, 20, 5]
    )[0]

    generated_types = []

    for i in range(total_cards):

        if i == 0:
            card_type = first_card

        else:
            card_type = random.choice(list(card_limits.keys()))

        generated_types.append(card_type)

        issue_date = fake.date_between(
            start_date="-8y",
            end_date="today"
        )

        expiry_date = issue_date + relativedelta(years=5)

        limit_min, limit_max = card_limits[card_type]

        cards.append({

            "Card_ID": f"CARD{card_id}",

            "Customer_ID": customer_id,

            "Card_Type": card_type,

            "Card_Network": random.choice(card_networks),

            "Credit_Limit": random.randint(
                limit_min,
                limit_max
            ),

            "Card_Status": random.choices(
                ["Active", "Blocked", "Expired", "Lost"],
                weights=[94, 2, 3, 1]
            )[0],

            "Contactless": random.choices(
                ["Yes", "No"],
                weights=[85, 15]
            )[0],

            "Card_Mode": random.choices(
                ["Physical", "Virtual"],
                weights=[80, 20]
            )[0],

            "Issue_Date": issue_date,

            "Expiry_Date": expiry_date

        })

        card_id += 1

cards_df = pd.DataFrame(cards)

print(cards_df.head())

print(f"Total Cards : {len(cards_df)}")

cards_df.to_csv("Cards_Data.csv", index=False)

print("Cards_Data.csv Saved Successfully")
```

```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# Set random seeds for consistency
np.random.seed(42)
random.seed(42)

print("Parent CSV files read kar rahe hain...")
customers_df = pd.read_csv('Cusmtomer_data.csv')
cards_df = pd.read_csv('Cards_Data.csv')
merchants_df = pd.read_csv('merchant_table.csv')

merchants_list = merchants_df.to_dict('records')
cust_map = customers_df.set_index('Customer_ID').to_dict('index')

# Category Wise Transaction Rules: (Min Amount, Max Amount, Typical Mean)
cat_amount_rules = {
    'Utilities': (300, 8000, 1500),
    'Education': (2000, 150000, 25000),
    'Travel': (1000, 60000, 12000),
    'Airline': (3000, 120000, 25000),
    'Fuel': (200, 5000, 1800),
    'Food Delivery': (100, 2500, 450),
    'Ecommerce': (200, 40000, 3500),
    'Hospital': (1000, 200000, 18000),
    'Electronics': (1500, 150000, 28000),
    'Hotel': (1500, 80000, 12000),
    'Pharmacy': (100, 10000, 1200),
    'Entertainment': (200, 8000, 1500),
    'Fashion': (500, 35000, 4500),
    'Grocery': (150, 12000, 2200),
    'Financial Services': (1000, 100000, 15000)
}

channels_by_mode = {
    'Physical': ['POS', 'ATM', 'POS', 'POS', 'Mobile App'],
    'Virtual': ['Online Web', 'Mobile App', 'Online Web', 'Mobile App']
}

def generate_transactions_large(n_rows=250000):
    tx_list = []
    
    active_cards = cards_df[cards_df['Card_Status'] == 'Active'].reset_index(drop=True)
    inactive_cards = cards_df[cards_df['Card_Status'] != 'Active'].reset_index(drop=True)
    
    active_card_records = active_cards.to_dict('records')
    inactive_card_records = inactive_cards.to_dict('records') if len(inactive_cards) > 0 else active_card_records
    
    start_dt = datetime(2023, 1, 1)
    end_dt = datetime(2026, 6, 30)
    
    print(f"Generating {n_rows:,} transactions... Please wait a few seconds.")
    
    for i in range(n_rows):
        txn_id = f"TXN{1000001 + i}"
        
        # 98% active cards, 2% attempts on inactive/blocked/expired cards
        if random.random() < 0.02:
            card_row = random.choice(inactive_card_records)
        else:
            card_row = random.choice(active_card_records)
            
        cust_id = card_row['Customer_ID']
        cust_info = cust_map[cust_id]
        merchant_row = random.choice(merchants_list)
        
        # Transaction Date Logic
        card_issue_dt = datetime.strptime(card_row['Issue_Date'], '%Y-%m-%d')
        card_expiry_dt = datetime.strptime(card_row['Expiry_Date'], '%Y-%m-%d')
        
        min_txn_dt = max(start_dt, card_issue_dt)
        max_txn_dt = min(end_dt, card_expiry_dt)
        
        if min_txn_dt >= max_txn_dt:
            txn_dt = card_issue_dt + timedelta(days=random.randint(1, 30))
        else:
            time_span = int((max_txn_dt - min_txn_dt).total_seconds())
            txn_dt = min_txn_dt + timedelta(seconds=random.randint(0, time_span))
            
        txn_date_str = txn_dt.strftime('%Y-%m-%d')
        txn_time_str = txn_dt.strftime('%H:%M:%S')
        
        # Channel & Device Mapping
        card_mode = card_row['Card_Mode']
        channel = random.choice(channels_by_mode[card_mode])
        if channel == 'POS':
            device_type = 'POS Terminal'
        elif channel == 'ATM':
            device_type = 'ATM Machine'
        elif channel == 'Mobile App':
            device_type = random.choice(['Android Mobile', 'iOS Mobile'])
        else:
            device_type = random.choice(['Windows PC', 'Mac Desktop', 'Android Mobile', 'iOS Mobile'])
            
        m_cat = merchant_row['Merchant_Category']
        card_status = card_row['Card_Status']
        min_amt, max_amt, mean_amt = cat_amount_rules.get(m_cat, (500, 10000, 3000))
        
        # Logic Flags Initializer
        is_fraud = 0
        fraud_reason = "None"
        status = "Successful"
        is_intl = 0
        
        # Fraud Logic Rules
        if card_status in ['Blocked', 'Expired', 'Lost']:
            is_fraud = 1
            fraud_reason = f"Transaction Attempt on {card_status} Card"
            status = "Declined"
            amt = round(random.uniform(min_amt, mean_amt * 1.5), 2)
            is_intl = 1 if (m_cat in ['Airline', 'Hotel', 'Travel'] and random.random() < 0.2) else 0
        else:
            if random.random() < 0.035:  # Controlled ~3.5% fraud on active cards
                is_fraud = 1
                fraud_type = random.choice(['HIGH_AMT', 'LATE_NIGHT_GEO', 'HIGH_RISK_MERCHANT', 'CROSS_BORDER', 'VELOCITY'])
                
                if fraud_type == 'HIGH_AMT':
                    credit_lim = card_row['Credit_Limit']
                    amt = round(random.uniform(credit_lim * 0.75, credit_lim * 1.3), 2)
                    fraud_reason = "Unusual High Transaction Amount vs Credit Limit"
                    status = random.choice(['Successful', 'Declined', 'Failed'])
                    is_intl = 1 if random.random() < 0.3 else 0
                    
                elif fraud_type == 'LATE_NIGHT_GEO':
                    amt = round(random.uniform(mean_amt * 1.5, max_amt), 2)
                    fraud_reason = "Late Night Geographical & Velocity Anomaly"
                    status = random.choice(['Successful', 'Declined'])
                    
                elif fraud_type == 'HIGH_RISK_MERCHANT':
                    amt = round(random.uniform(mean_amt * 2, max_amt * 1.1), 2)
                    fraud_reason = "High Risk Merchant Category Spike"
                    status = random.choice(['Successful', 'Failed'])
                    
                elif fraud_type == 'CROSS_BORDER':
                    is_intl = 1
                    amt = round(random.uniform(mean_amt * 2.5, max_amt), 2)
                    fraud_reason = "Unusual Cross-Border / International Transaction"
                    status = random.choice(['Successful', 'Declined'])
                    
                elif fraud_type == 'VELOCITY':
                    amt = round(random.uniform(min_amt, mean_amt * 2), 2)
                    fraud_reason = "Rapid Velocity & Multiple Failed Auth Attempts"
                    status = "Failed"
            else:
                # Normal Transactions
                is_fraud = 0
                fraud_reason = "None"
                
                if m_cat in ['Airline', 'Hotel', 'Travel'] and random.random() < 0.12:
                    is_intl = 1
                elif random.random() < 0.015:
                    is_intl = 1
                    
                scale_val = (max_amt - min_amt) / 4.0
                amt = round(np.random.normal(loc=mean_amt, scale=scale_val), 2)
                amt = max(min_amt, min(amt, max_amt))
                
                st_rand = random.random()
                if st_rand < 0.95:
                    status = "Successful"
                elif st_rand < 0.98:
                    status = "Failed"
                else:
                    status = "Declined"
                    
        # Payment Method Mapping
        if channel == 'POS':
            pay_method = random.choice(['Credit Card', 'Debit Card'])
        elif channel == 'ATM':
            pay_method = 'Debit Card'
        elif channel == 'Mobile App':
            pay_method = random.choice(['UPI', 'Credit Card', 'Debit Card'])
        else:
            pay_method = random.choice(['Credit Card', 'Debit Card', 'Net Banking', 'UPI'])

        tx_list.append({
            'Transaction_ID': txn_id,
            'Customer_ID': cust_id,
            'Card_ID': card_row['Card_ID'],
            'Merchant_ID': merchant_row['Merchant_ID'],
            'Transaction_Date': txn_date_str,
            'Transaction_Time': txn_time_str,
            'Transaction_Amount': amt,
            'Payment_Method': pay_method,
            'Transaction_Channel': channel,
            'Device_Type': device_type,
            'Transaction_Status': status,
            'Is_International': is_intl,
            'Fraud_Flag': is_fraud,
            'Fraud_Reason': fraud_reason,
            'Merchant_Risk_Level': merchant_row['Merchant_Risk_Level'],
            'Merchant_Category': merchant_row['Merchant_Category'],
            'Customer_State': cust_info['State'],
            'Customer_City': cust_info['City'],
            'Merchant_State': merchant_row['State'],
            'Merchant_City': merchant_row['City']
        })
        
    return pd.DataFrame(tx_list)

# Generate 250,000 Rows
df_tx = generate_transactions_large(250000)

# Save to CSV
output_filename = 'Transaction_Data_250k.csv'
df_tx.to_csv(output_filename, index=False)
print(f"SUCCESS: {len(df_tx):,} rows successfully created and saved in '{output_filename}'!")
```
